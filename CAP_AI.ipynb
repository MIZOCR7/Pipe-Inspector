```python
import zipfile
import os

zip_path = "/content/YOLO.v3i.yolov8.zip"

with zipfile.ZipFile(zip_path, "r") as zip_ref:
    zip_ref.extractall("/content/")

print(os.listdir("/content/"))
```

```python
from ultralytics import YOLO

model = YOLO("yolo26s.pt")

results = model.train(
    data="/content/data.yaml",
    epochs=100,
    imgsz=640,
    batch=16,
    patience=20,
    lr0=0.01
)
```
